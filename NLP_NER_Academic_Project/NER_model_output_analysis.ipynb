```python
import spacy
import os
if not os.path.exists("en_core_web_sm"):
    os.system("python -m spacy download en_core_web_sm")
nlp = spacy.load("en_core_web_sm")
```

```python
doc = nlp("Apple CEO Tim Cook visited India")
```

```python
for ent in doc.ents:
    print(ent.text, ent.label_)
```
