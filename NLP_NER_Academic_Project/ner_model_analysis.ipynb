```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
sentences = [
    # News
    "Elon Musk launched a new rocket in Texas",
    "Google announced a new AI feature today",
    "Microsoft opened a new office in London",

    # Informal / social media
    "elon musk tweeted lol 😂",
    "amazon sale is crazy bro",
    "tesla stock going up!!",

    # Ambiguous
    "Apple is very tasty",
    "Apple released a new iPhone",
    
    # Mixed
    "Virat Kohli played in Mumbai",
    "I met Rahul yesterday",
    "Meta is working on virtual reality",
    "Sundar Pichai visited India"
]
```

```python
for text in sentences:
    print("\nSentence:", text)
    doc = nlp(text)

    for ent in doc.ents:
        print(ent.text, ent.label_)
```

```python
for text in sentences:
    print("\nSentence:", text)
    
    doc = nlp(text)
    
    for ent in doc.ents:
        print(ent.text, ent.label_)
```

```python
import json

results = []

for text in sentences:
    doc = nlp(text)
    ents = [(ent.text, ent.label_) for ent in doc.ents]

    results.append({
        "sentence": text,
        "entities": ents
    })

with open("spacy_ner_results.json", "w") as f:
    json.dump(results, f, indent=4)
```
