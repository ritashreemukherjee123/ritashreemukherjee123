```python
import os

base_path = "conll2003"
```

```python
def load_conll(file_path):
    sentences = []
    labels = []
    sentence = []
    tag = []

    with open(file_path, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()

            if line == "":
                if sentence:
                    # skip DOCSTART sentences
                    if sentence != ["-DOCSTART-"]:
                        sentences.append(sentence)
                        labels.append(tag)
                    sentence = []
                    tag = []
            else:
                parts = line.split()
                word = parts[0]
                ner = parts[-1]

                sentence.append(word)
                tag.append(ner)

    return sentences, labels
```

```python
train_sentences, train_labels = load_conll(os.path.join(base_path, "eng.train"))
dev_sentences, dev_labels = load_conll(os.path.join(base_path, "eng.testa"))
test_sentences, test_labels = load_conll(os.path.join(base_path, "eng.testb"))

print(len(train_sentences))
print(train_sentences[0], train_labels[0])
```

CREATE tag2idx AND idx2tag

```python
# Create tag mappings
unique_tags = sorted(set(tag for sent in train_labels for tag in sent))

tag2idx = {tag: idx for idx, tag in enumerate(unique_tags)}
idx2tag = {idx: tag for tag, idx in tag2idx.items()}

print(tag2idx)
```

LOAD TOKENIZER (WORDPIECE)

```python
from transformers import BertTokenizerFast

tokenizer = BertTokenizerFast.from_pretrained("bert-base-cased")
```

CREATE PYTORCH DATASET CLASS

```python
import torch
from torch.utils.data import Dataset

class NERDataset(Dataset):
    def __init__(self, sentences, labels):
        self.sentences = sentences
        self.labels = labels

    def __len__(self):
        return len(self.sentences)

    def __getitem__(self, idx):
        words = self.sentences[idx]
        tags = self.labels[idx]

        encoding = tokenizer(
            words,
            is_split_into_words=True,
            padding="max_length",
            truncation=True,
            max_length=128,
            return_tensors="pt"
        )

        label_ids = []
        word_ids = encoding.word_ids()
        prev_word = None

        for word_id in word_ids:
            if word_id is None:
                label_ids.append(-100)     # ignore padding
            elif word_id != prev_word:
                label_ids.append(tag2idx[tags[word_id]])
            else:
                label_ids.append(-100)     # ignore subword
            prev_word = word_id

        return {
            "input_ids": encoding["input_ids"].squeeze(),
            "attention_mask": encoding["attention_mask"].squeeze(),
            "labels": torch.tensor(label_ids)
        }
```

CREATE TRAIN & DEV DATASET

```python
train_dataset = NERDataset(
    train_sentences[:12000],
    train_labels[:12000]
)

dev_dataset = NERDataset(
    dev_sentences[:3000],
    dev_labels[:3000]
)
```

```python
from transformers import BertForTokenClassification

model = BertForTokenClassification.from_pretrained(
    "bert-base-cased",
    num_labels=len(tag2idx),
    id2label=idx2tag,
    label2id=tag2idx
)
```

STEP 6: TRAINING ARGUMENTS

```python
from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir="./bert-ner-trained",
    num_train_epochs=6,          # enough learning
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    learning_rate=3e-5,
    eval_strategy="epoch",
    save_strategy="epoch",
    logging_steps=200,
    report_to="none"
)
```

```python
from transformers import Trainer
from seqeval.metrics import f1_score

def compute_metrics(p):
    preds, labels = p
    preds = preds.argmax(axis=2)

    true_preds = [
        [idx2tag[p] for p, l in zip(pred, lab) if l != -100]
        for pred, lab in zip(preds, labels)
    ]
    true_labels = [
        [idx2tag[l] for p, l in zip(pred, lab) if l != -100]
        for pred, lab in zip(preds, labels)
    ]

    return {"f1": f1_score(true_labels, true_preds)}

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=dev_dataset,
    tokenizer=tokenizer,
    compute_metrics=compute_metrics
)

trainer.train()
```

```python
model.save_pretrained("./bert-ner-trained")
tokenizer.save_pretrained("./bert-ner-trained")

print("Model and tokenizer saved successfully!")
```

```python
import torch

def test_model(sentence):
    sentence = sentence.title()
    words = sentence.split()

    enc = tokenizer(words, is_split_into_words=True, return_tensors="pt")
    with torch.no_grad():
        output = model(**enc)

    preds = torch.argmax(output.logits, dim=2)[0].tolist()
    word_ids = enc.word_ids()

    results = []
    for token, wid, pred in zip(enc.tokens(), word_ids, preds):
        if wid is not None:
            results.append((token, idx2tag[pred]))

    return results
```

```python
def merge_subwords(predictions):
    merged = []
    current_word = ""
    current_tag = ""

    for token, tag in predictions:
        if token.startswith("##"):
            current_word += token[2:]
        else:
            if current_word:
                merged.append((current_word, current_tag))
            current_word = token
            current_tag = tag

    if current_word:
        merged.append((current_word, current_tag))

    return merged
```

```python
raw_output = test_model("Elon Musk founded SpaceX")
print("Raw output:")
print(raw_output)

print("\nMerged output:")
print(merge_subwords(raw_output))
```

```python
print(test_model("Elon Musk founded SpaceX"))
print(test_model("EU rejects German call to boycott British lamb"))
print(test_model("BTS performed at the award ceremony"))
```
